## Project Description

This project calculates:
- Total marks
- Percentage
- Student grade
- Pass or fail status

## Technologies Used
- Python
- Jupyter Notebook

```python
# Student Grade Calculator

print("Welcome to Student Grade Calculator")

# Get student name
name = input("Enter your name: ")

# Enter subject marks
maths = float(input("Maths marks: "))
physics = float(input("Physics marks: "))
programming = float(input("Programming marks: "))
english = float(input("English marks: "))

# Calculate total and percentage
total = maths + physics + programming + english
percentage = (total / 400) * 100

# Find student grade
if percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

# Display final result
print("\nStudent Name:", name)
print("Total Marks:", total)
print("Percentage:", round(percentage, 2), "%")
print("Final Grade:", grade)
```
